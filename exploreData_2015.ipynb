```python
import pandas as pd
import numpy as np

raw = pd.read_csv('2015.csv')
df = raw.copy()
df.info()
pd.set_option('display.max_columns', None)
df.head()
```

```python
# counts how many times each distinct value appears
df['DIABETE3'].value_counts(dropna=False)

# Codes: 1 = Yes, 2 = Yes but only during pregnancy, 3 = No, 4 = Pre-diabetes, 7 = Don't know, 9 = Refused, NaN = missing
```

```python
df = df[df['DIABETE3'].isin([1, 2, 3, 4])]
df['target'] = df['DIABETE3'].map({1: 2, 4: 1, 3: 0, 2: 0})   # example: 3 classes
```

```python
# drop all colums with no medical meaning
missing = df.isna().mean()
keep = missing[missing < 0.5].index
df = df[keep]
```
